```python
import pandas as pd

df = pd.read_csv("cleaned_data.csv", parse_dates=["time"])
df["year"] = df["time"].dt.year
df["is_weekend"] = df["is_weekend"] == 1      

pollutants = ["pm25", "pm10", "co", "no2", "so2", "o3", "aqi"]
weather = ["temperature", "humidity", "rainfall", "pressure", "wind_speed"]   # wind_direction removed
pct_cols = ["pm25_pct", "pm10_pct", "no2_pct", "so2_pct", "o3_pct", "co_pct"]


print("1. City averages")
print(df.groupby("city")[weather + pollutants].mean().round(2))


print("\n2. Region averages")
print(df.groupby("region")[weather + pollutants].mean().round(2))


print("\n3. Season averages")
print(df.groupby("season")[pollutants].mean().round(2))


print("\n4. Month averages")
print(df.groupby("month")[pollutants].mean().round(2))


print("\n5. Year averages")
print(df.groupby("year")[pollutants].mean().round(2))

 
print("\n5b. Nov-Dec averages by year")
print(df[df["month"] >= 11].groupby("year")[["pm25", "pm10", "no2", "aqi"]].mean().round(2))


print("\n6. AQI: city vs season")
print(df.pivot_table(index="city", columns="season", values="aqi").round(2))


print("\n7. AQI: city vs month")
print(df.pivot_table(index="city", columns="month", values="aqi").round(2))


print("\n8. Weekend (True) vs weekday (False)")
print(df.groupby("is_weekend")[pollutants].mean().round(2))


print("\n9. AQI category count")
print(pd.crosstab(df["city"], df["aqi_category"]))


print("\n10. % of days above safe limit")
print(((df[pct_cols] > 100).groupby(df["city"]).mean() * 100).round(1))


cols = weather + pollutants
diff = df[cols] - df.groupby("city")[cols].transform("mean")
print("\n11. Correlation (-1 to +1)")
print(diff.corr().round(2))

df["rainy"] = df["rainfall"] > 0
print("\n12. Rainy (True) vs dry (False) days")
print(df.groupby("rainy")[pollutants].mean().round(2))


print("\n13. Worst AQI day per city")
print(df.loc[df.groupby("city")["aqi"].idxmax(), ["city", "time", "aqi", "pm25", "pm10"]])
```
